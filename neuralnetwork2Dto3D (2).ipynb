```python
# =========================
# Cell 1: Estimate focal length from two training files
# File 1: x_pixel, y_pixel, z_real
# File 2: x_real, y_real
# =========================

import pandas as pd
import numpy as np
from google.colab import files

# -------------------------
# 1) Upload pixel + z file
# -------------------------
print("Upload file 1: x_pixel, y_pixel, z_real")
uploaded_2d = files.upload()
FILE_2D_PATH = list(uploaded_2d.keys())[0]

df_2d = pd.read_excel(FILE_2D_PATH)

print("File 1 uploaded:", FILE_2D_PATH)
print("File 1 columns:")
print(df_2d.columns.tolist())
display(df_2d.head())


# -------------------------
# 2) Upload real x,y file
# -------------------------
print("Upload file 2: x_real, y_real")
uploaded_real = files.upload()
FILE_REAL_PATH = list(uploaded_real.keys())[0]

df_real = pd.read_excel(FILE_REAL_PATH)

print("File 2 uploaded:", FILE_REAL_PATH)
print("File 2 columns:")
print(df_real.columns.tolist())
display(df_real.head())


# -------------------------
# 3) Select columns by position
# چون اسم ستون‌ها متفاوت است، فعلاً با شماره ستون می‌خوانیم
# فایل اول باید حداقل 3 ستون داشته باشد:
# col 0 = x_pixel
# col 1 = y_pixel
# col 2 = z_real
#
# فایل دوم باید حداقل 2 ستون داشته باشد:
# col 0 = x_real
# col 1 = y_real
# -------------------------

if df_2d.shape[1] < 3:
    raise ValueError("File 1 must have at least 3 columns: x_pixel, y_pixel, z_real")

if df_real.shape[1] < 2:
    raise ValueError("File 2 must have at least 2 columns: x_real, y_real")

x_pixel = df_2d.iloc[:, 0].astype(float).values
y_pixel = df_2d.iloc[:, 1].astype(float).values
z_real  = df_2d.iloc[:, 2].astype(float).values

x_real = df_real.iloc[:, 0].astype(float).values
y_real = df_real.iloc[:, 1].astype(float).values


# -------------------------
# 4) Check row counts
# -------------------------

n = min(len(x_pixel), len(y_pixel), len(z_real), len(x_real), len(y_real))

x_pixel = x_pixel[:n]
y_pixel = y_pixel[:n]
z_real  = z_real[:n]
x_real  = x_real[:n]
y_real  = y_real[:n]

print("Number of paired samples:", n)

# حذف ردیف‌های خراب
valid_mask = (
    np.isfinite(x_pixel) &
    np.isfinite(y_pixel) &
    np.isfinite(z_real) &
    np.isfinite(x_real) &
    np.isfinite(y_real) &
    (z_real != 0)
)

x_pixel = x_pixel[valid_mask]
y_pixel = y_pixel[valid_mask]
z_real  = z_real[valid_mask]
x_real  = x_real[valid_mask]
y_real  = y_real[valid_mask]

print("Valid samples after cleaning:", len(z_real))


# -------------------------
# 5) Estimate camera parameters
# مدل:
# x_pixel = fx * (x_real / z_real) + cx
# y_pixel = fy * (y_real / z_real) + cy
# -------------------------

x_norm = x_real / z_real
y_norm = y_real / z_real

A_x = np.column_stack([x_norm, np.ones_like(x_norm)])
fx, cx = np.linalg.lstsq(A_x, x_pixel, rcond=None)[0]

A_y = np.column_stack([y_norm, np.ones_like(y_norm)])
fy, cy = np.linalg.lstsq(A_y, y_pixel, rcond=None)[0]

print("Estimated camera parameters:")
print("fx =", fx)
print("fy =", fy)
print("cx =", cx)
print("cy =", cy)

# برای محاسبه فاصله از روی ارتفاع، focal عمودی مهم‌تر است
FOCAL_FOR_DISTANCE = fy

print("FOCAL_FOR_DISTANCE =", FOCAL_FOR_DISTANCE)


# -------------------------
# 6) Error check
# -------------------------

x_pixel_pred = fx * x_norm + cx
y_pixel_pred = fy * y_norm + cy

x_error = x_pixel - x_pixel_pred
y_error = y_pixel - y_pixel_pred

print("Mean absolute x error:", np.mean(np.abs(x_error)))
print("Mean absolute y error:", np.mean(np.abs(y_error)))
print("Median absolute x error:", np.median(np.abs(x_error)))
print("Median absolute y error:", np.median(np.abs(y_error)))


# -------------------------
# 7) Save focal parameters
# -------------------------

camera_params_df = pd.DataFrame({
    "parameter": ["fx", "fy", "cx", "cy", "FOCAL_FOR_DISTANCE"],
    "value": [fx, fy, cx, cy, FOCAL_FOR_DISTANCE]
})

CAMERA_PARAMS_PATH = "estimated_camera_parameters.xlsx"
camera_params_df.to_excel(CAMERA_PARAMS_PATH, index=False)

print("Saved camera parameters:", CAMERA_PARAMS_PATH)
files.download(CAMERA_PARAMS_PATH)
```

```python
# =========================
# Train two separate fusion networks for X and Y
# No cleaning, no column removal
# Input files:
#   column 1 = audio estimate
#   column 2 = visual/image estimate
#
# Output files:
#   column 1 = target / true value
#
# Zero rule:
#   if one input column is zero, replace it with the other one
#   if both are zero, keep both zero
# =========================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error


# =========================
# 1) مسیر فایل‌های آموزشی
# =========================

X_INPUT_TRAIN_PATH  = r'/content/drive/MyDrive/lipmovement/xinputtrain.xlsx'
X_OUTPUT_TRAIN_PATH = r'/content/drive/MyDrive/lipmovement/xoutputtrain.xlsx'

Y_INPUT_TRAIN_PATH  = r'/content/drive/MyDrive/lipmovement/yinputtrain.xlsx'
Y_OUTPUT_TRAIN_PATH = r'/content/drive/MyDrive/lipmovement/youtputtrain.xlsx'

MODEL_OUTPUT_DIR = r'/content/drive/MyDrive/lipmovement/fusion_models_xy'
os.makedirs(MODEL_OUTPUT_DIR, exist_ok=True)


# =========================
# 2) قانون صفر
# =========================

def apply_zero_rule_array(X):
    """
    X must have 2 columns:
    col 0 = audio estimate
    col 1 = visual estimate
    """

    X_fixed = X.copy().astype(float)

    audio = X_fixed[:, 0]
    visual = X_fixed[:, 1]

    # اگر audio صفر بود و visual غیرصفر بود
    mask_audio_zero = (audio == 0) & (visual != 0)
    X_fixed[mask_audio_zero, 0] = visual[mask_audio_zero]

    # اگر visual صفر بود و audio غیرصفر بود
    mask_visual_zero = (visual == 0) & (audio != 0)
    X_fixed[mask_visual_zero, 1] = audio[mask_visual_zero]

    # اگر هر دو صفر باشند، همان صفر می‌مانند

    return X_fixed


# =========================
# 3) ساخت مدل
# =========================

def build_model():
    model = Sequential()
    model.add(Dense(16, input_dim=2, kernel_initializer='normal', activation='relu'))
    model.add(Dense(1))
    model.compile(loss='mse', optimizer='adam')
    return model


# =========================
# 4) آموزش یک محور
# =========================

def train_axis_model(input_path, output_path, axis_name):

    print("\n" + "="*70)
    print(f"Training {axis_name.upper()} fusion model")
    print("="*70)

    # خواندن مستقیم فایل‌ها، بدون پاکسازی
    input_df = pd.read_excel(input_path)
    output_df = pd.read_excel(output_path)

    print(f"{axis_name.upper()} input shape:", input_df.shape)
    print(f"{axis_name.upper()} output shape:", output_df.shape)

    print(f"{axis_name.upper()} input columns:", input_df.columns.tolist())
    print(f"{axis_name.upper()} output columns:", output_df.columns.tolist())

    print("Input preview:")
    print(input_df.head(5).to_string())

    print("Output preview:")
    print(output_df.head(5).to_string())

    # ورودی فقط دو ستون اول
    X = input_df.iloc[:, 0:2].astype(float).values

    # خروجی فقط ستون اول
    Y = output_df.iloc[:, 0:1].astype(float).values

    # اعمال قانون صفر روی ورودی
    X = apply_zero_rule_array(X)

    # برابر کردن تعداد ردیف‌ها
    n = min(len(X), len(Y))
    X = X[:n]
    Y = Y[:n]

    print(f"{axis_name.upper()} final X shape:", X.shape)
    print(f"{axis_name.upper()} final Y shape:", Y.shape)

    # ذخیره ورودی اصلاح‌شده با قانون صفر برای کنترل
    fixed_input_df = pd.DataFrame(X, columns=[
        f"{axis_name}_audio_fixed",
        f"{axis_name}_visual_fixed"
    ])

    fixed_input_path = os.path.join(
        MODEL_OUTPUT_DIR,
        f"{axis_name}_inputtrain_zero_rule_applied.xlsx"
    )

    fixed_input_df.to_excel(fixed_input_path, index=False)

    print("Zero-rule input saved:", fixed_input_path)

    # train/test split
    xtrain, xtest, ytrain, ytest = train_test_split(
        X,
        Y,
        test_size=0.1,
        random_state=42
    )

    print("xtrain:", xtrain.shape)
    print("xtest :", xtest.shape)
    print("ytrain:", ytrain.shape)
    print("ytest :", ytest.shape)

    # ساخت مدل
    model = build_model()
    model.summary()

    # آموزش
    history = model.fit(
        xtrain,
        ytrain,
        validation_split=0.1,
        epochs=100,
        batch_size=8,
        verbose=1
    )

    # ارزیابی
    ypred_test = model.predict(xtest)

    mse = mean_squared_error(ytest[:, 0], ypred_test[:, 0])
    mae = mean_absolute_error(ytest[:, 0], ypred_test[:, 0])

    print(f"{axis_name.upper()} Test MSE:", mse)
    print(f"{axis_name.upper()} Test MAE:", mae)

    # نمودار loss
    plt.figure(figsize=(10, 6))
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title(f'{axis_name.upper()} Fusion Model Loss')
    plt.xlabel('Epoch')
    plt.ylabel('MSE')
    plt.legend()
    plt.grid(True)
    plt.show()

    # ذخیره مدل
    model_json_path = os.path.join(MODEL_OUTPUT_DIR, f'{axis_name}_fusion_model.json')
    model_weights_path = os.path.join(MODEL_OUTPUT_DIR, f'{axis_name}_fusion_model.weights.h5')

    with open(model_json_path, "w") as json_file:
        json_file.write(model.to_json())

    model.save_weights(model_weights_path)

    print(f"Saved {axis_name.upper()} model JSON:", model_json_path)
    print(f"Saved {axis_name.upper()} model weights:", model_weights_path)

    return model, history, mse, mae


# =========================
# 5) آموزش مدل X و مدل Y
# =========================

x_model, x_history, x_mse, x_mae = train_axis_model(
    X_INPUT_TRAIN_PATH,
    X_OUTPUT_TRAIN_PATH,
    axis_name="x"
)

y_model, y_history, y_mse, y_mae = train_axis_model(
    Y_INPUT_TRAIN_PATH,
    Y_OUTPUT_TRAIN_PATH,
    axis_name="y"
)


# =========================
# 6) خلاصه نهایی
# =========================

print("\n" + "="*70)
print("Both models trained successfully.")
print("="*70)

print("X model:")
print("MSE:", x_mse)
print("MAE:", x_mae)

print("\nY model:")
print("MSE:", y_mse)
print("MAE:", y_mae)

print("\nModels saved in:")
print(MODEL_OUTPUT_DIR)
```

```python
# =========================
# Cell 2: Predict Audiovisual result using trained X and Y fusion models
#
# Required trained models from previous cell:
#   x_model
#   y_model
#
# Input X Excel:
#   column 1 = x estimate from audio
#   column 2 = x estimate from image/visual
#
# Input Y Excel:
#   column 1 = y estimate from audio
#   column 2 = y estimate from image/visual
#
# Output:
#   x_audiovisual_pred
#   y_audiovisual_pred
# =========================

import os
import numpy as np
import pandas as pd
from google.colab import files


# =========================
# 1) Zero rule
# =========================

def apply_zero_rule_array(X):
    """
    X must have exactly 2 columns:
    col 0 = audio estimate
    col 1 = visual estimate

    Rule:
    if one column is zero, replace it with the other one
    if both are zero, keep both zero
    """

    X_fixed = X.copy().astype(float)

    audio = X_fixed[:, 0]
    visual = X_fixed[:, 1]

    # audio = 0, visual != 0
    mask_audio_zero = (audio == 0) & (visual != 0)
    X_fixed[mask_audio_zero, 0] = visual[mask_audio_zero]

    # visual = 0, audio != 0
    mask_visual_zero = (visual == 0) & (audio != 0)
    X_fixed[mask_visual_zero, 1] = audio[mask_visual_zero]

    # both zero -> unchanged

    return X_fixed


# =========================
# 2) Read Excel without cleaning
# =========================

def read_two_column_excel(path):
    df = pd.read_excel(path)

    print("Reading:", path)
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print(df.head(5).to_string())

    # فقط دو ستون اول
    X = df.iloc[:, 0:2].astype(float).values

    if X.shape[1] != 2:
        raise ValueError("Input Excel must have exactly 2 usable columns.")

    X = apply_zero_rule_array(X)

    return df, X


# =========================
# 3) Upload X and Y input files
# =========================

print("Upload X input Excel file: x_audio_estimate, x_visual_estimate")
uploaded_x = files.upload()
X_INPUT_PATH = list(uploaded_x.keys())[0]

print("Upload Y input Excel file: y_audio_estimate, y_visual_estimate")
uploaded_y = files.upload()
Y_INPUT_PATH = list(uploaded_y.keys())[0]


# =========================
# 4) Read X and Y files
# =========================

x_df_original, X_new = read_two_column_excel(X_INPUT_PATH)
y_df_original, Y_new = read_two_column_excel(Y_INPUT_PATH)

print("X_new shape:", X_new.shape)
print("Y_new shape:", Y_new.shape)


# =========================
# 5) Equalize number of rows
# =========================

n = min(len(X_new), len(Y_new))

X_new = X_new[:n]
Y_new = Y_new[:n]

print("Final paired rows:", n)


# =========================
# 6) Predict with trained models
# =========================

x_pred = x_model.predict(X_new)
y_pred = y_model.predict(Y_new)


# =========================
# 7) Build final audiovisual output
# =========================

result_df = pd.DataFrame({
    "frame_index": np.arange(1, n + 1),
    "x_audiovisual_pred": x_pred[:, 0],
    "y_audiovisual_pred": y_pred[:, 0]
})

print("Audiovisual output preview:")
print(result_df.head(20).to_string())


# =========================
# 8) Save output
# =========================

OUTPUT_DIR = r'/content/drive/MyDrive/lipmovement/audiovisual_outputs'
os.makedirs(OUTPUT_DIR, exist_ok=True)

x_base = os.path.splitext(os.path.basename(X_INPUT_PATH))[0]
y_base = os.path.splitext(os.path.basename(Y_INPUT_PATH))[0]

OUTPUT_PATH = os.path.join(
    OUTPUT_DIR,
    f"audiovisual_result_{x_base}_{y_base}.xlsx"
)

result_df.to_excel(OUTPUT_PATH, index=False)

print("Saved audiovisual result:")
print(OUTPUT_PATH)

files.download(OUTPUT_PATH)
```
